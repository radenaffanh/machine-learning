```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os

folder_path = r"D:\Documents\5th Term RAIE 24'\Machine Learning Lab (RK-A1)\TM 4 Modul 3\Preprocessing (1)\Preprocessing\pidato"

# Ambil nama semua file di dalam folder
FileNames = os.listdir(folder_path)

print(FileNames)
```

```python
pidato_df = pd.DataFrame(
    index=range(len(FileNames)),
    columns=['File Name', 'The Content']
)

for i, f_name in enumerate(FileNames):

    file_path = os.path.join(folder_path, f_name)

    with open(file_path, "r", encoding="utf-8") as f:
        f_content = f.read()

    pidato_df.at[i, 'File Name'] = f_name
    pidato_df.at[i, 'The Content'] = f_content

pidato_df.columns = ['FileName', 'Content']

print(pidato_df)
```

Contoh 4: Membongkar Table

```python
#Unpacking FileName 
Months = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Oct','Sep','Nov','Dec'] 
def SeparateCity(v): 
    for mon in Months: 
        if (mon in v): 
            return v[:v.find(mon)] 
pidato_df['City'] = pidato_df.FileName.apply(SeparateCity) 
 
def SeparateDate(r): 
    return r.FileName[len(r.City):r.FileName.find('.txt')] 
 
pidato_df['Date'] = pidato_df.apply(SeparateDate,axis=1) 
pidato_df.Date = pd.to_datetime(pidato_df.Date,format='%b%d_%Y') 
 
def extractDMY(r): 
    r['Month'] = r.Date.month 
    r['Year'] = r.Date.year 
    return r 
pidato_df = pidato_df.apply(extractDMY,axis=1) 
pidato_df.drop(columns=['FileName'],inplace=True) 
pidato_df.head()
```

```python
#Unpacking content 
Words = ['vote','tax','campaign','economy'] 
 
def FindWordRatio(row): 
    total_n_words = len(row.Content.split(' ')) 
    for w in Words: 
        row['r_{}'.format(w)] = row.Content.count(w)/total_n_words 
    return row 
 
pidato_df = pidato_df.apply(FindWordRatio,axis=1) 
pidato_df.head()
```

Contoh 5: Restrukturisasi tabel 

```python
customer_df = pd.read_csv(r"D:\Documents\5th Term RAIE 24'\Machine Learning Lab (RK-A1)\TM 3 Modul 2\Preprocessing (1)\Preprocessing\Customer Churn.csv")
customer_df.head(1) 
customer_df.columns = ['Call_Failure', 'Complains', 
        'Subscription_Length', 'Seconds_of_Use', 
        'Frequency_of_use', 'Frequency_of_SMS', 
        'Distinct_Called_Numbers','Status', 'Churn'] 
churn_possibilities = customer_df.Churn.unique() 
box_sr = pd.Series('',index = churn_possibilities) 
for poss in churn_possibilities: 
    BM = customer_df.Churn == poss 
    box_sr[poss] = customer_df[BM].Call_Failure.values 
#Plot data origin 
print(box_sr) 
plt.boxplot(box_sr,vert=False) 
plt.yticks([1,2],['Not Churn','Churn']) 
plt.show()
```

```python
#Restrukturisasi data 
select_columns = ['Call_Failure','Subscription_Length','Seconds_of_Use','Frequency_of_use', 
                  'Frequency_of_SMS','Distinct_Called_Numbers'] 
churn_possibilities = customer_df.Churn.unique() 
plt.figure(figsize=(15,5)) 
for i,sc in enumerate(select_columns): 
    for poss in churn_possibilities: 
        BM = customer_df.Churn == poss 
        box_sr[poss] = customer_df[BM][sc].values 
    plt.subplot(2,3,i+1) 
    plt.boxplot(box_sr,vert=False) 
    plt.yticks([1,2],['Not Churn','Churn']) 
    plt.title(sc) 
plt.tight_layout() 
plt.show()
```

Contoh 16: Standarisasi Data 

```python
from sklearn.preprocessing import StandardScaler
import pandas
import numpy

# Open data set
url= r"D:\Documents\5th Term RAIE 24'\Machine Learning Lab (RK-A1)\TM 4 Modul 3\Preprocessing (1)\Preprocessing\data.csv"

# Data parameters
names = ['Duration', 'Date', 'Pulse', 'Maxpulse', 'Calories']

# Preparation dataframe
dataframe = pandas.read_csv(
    url,
    header=None,
    names=names,
    quoting=3,
    on_bad_lines='skip'
)

# Membersihkan kolom Calories agar numerik
calories = dataframe['Calories'].astype(str).str.replace('"', '', regex=False).str.strip()
calories = pandas.to_numeric(calories, errors='coerce')

# Menangani missing value
calories = calories.fillna(calories.mean())

# Mengubah menjadi array 2 dimensi
calories = calories.to_numpy().reshape(-1, 1)

# Standardize data
scaler = StandardScaler()
rescaledX = scaler.fit_transform(calories)

# Menampilkan hasil
print("Standardized Data (First 5 Rows):\n", rescaledX[:5])

print("\nStandardized Mean:", rescaledX.mean())
print("Standard deviation after standardization:", rescaledX.std())
```
