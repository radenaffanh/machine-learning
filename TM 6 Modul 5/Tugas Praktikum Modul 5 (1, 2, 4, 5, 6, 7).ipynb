```python
def permintaan_turun(x): 
    if(x<=1000): 
        nilai = 1 
    elif(x>1000 and x<5000): 
        nilai = (5000-x)/(4000) 
    elif(x>=5000): 
        nilai = 0
    return nilai 
 
def permintaan_naik(x): 
    if(x<=1000): 
        nilai = 0 
    elif(x>1000 and x<5000): 
        nilai = (x-1000)/(4000) 
    elif(x>=5000): 
        nilai = 1 
 
    return nilai 
 
def persediaan_sedikit(x): 
    if(x<=100): 
        nilai = 1 
    elif(x>100 and x<600): 
        nilai = (600-x)/(500) 
    elif(x>=600): 
        nilai = 0 
 
    return nilai 
 
def persediaan_banyak(x): 
    if(x<=100): 
        nilai = 0 
    elif(x>100 and x<600): 
        nilai = (x-100)/(500) 
    elif(x>=600): 
        nilai = 1 
 
    return nilai 
 
def produksi_berkurang(x): 
    if(x<=2000): 
        nilai = 1 
    elif(x>2000 and x<7000): 
        nilai = (7000-x)/(5000) 
    elif(x>=7000): 
        nilai = 0 
 
    return nilai 
 
def produksi_bertambah(x): 
    if(x<=2000): 
        nilai = 0 
    elif(x>2000 and x<7000): 
        nilai = (x-2000)/(5000) 
    elif(x>=7000): 
        nilai = 1 
 
    return nilai 
 
def rule1(x,y): 
    if (permintaan_turun(x)<=persediaan_banyak(y)): 
        alpha = permintaan_turun(x) 
    else:
        alpha = persediaan_banyak(y)
        z = 7000-(5000 * alpha) 
    return alpha,z; 
 
def rule2(x,y): 
  if (permintaan_turun(x)<=persediaan_sedikit(y)): 
    alpha = permintaan_turun(x) 
  else: 
    alpha = persediaan_sedikit(y) 
 
  z = 7000-(5000 * alpha) 
  return alpha,z; 
 
def rule3(x,y): 
  if (permintaan_naik(x)<=persediaan_banyak(y)): 
    alpha = permintaan_naik(x) 
  else: 
    alpha = persediaan_banyak(y) 
 
  z = 2000+(5000 * alpha) 
  return alpha,z; 
 
def rule4(x,y): 
  if (permintaan_naik(x)<=persediaan_sedikit(y)): 
    alpha = permintaan_naik(x) 
  else: 
    alpha = persediaan_sedikit(y) 
 
  z = 2000+(5000 * alpha) 
  return alpha,z; 
 
def tsukamoto(x,y): 
  alpha1,z1=rule1(x,y) 
  alpha2,z2=rule2(x,y) 
  alpha3,z3=rule3(x,y) 
  alpha4,z4=rule4(x,y) 
  alpha=(alpha1*z1)+(alpha2*z2)+(alpha3*z3)+(alpha4*z4) 
  z=alpha1+alpha2+alpha3+alpha4 
  return alpha/z 
 
p=4000 
q=300 
 
def rule1(x, y):
    alpha = min(permintaan_turun(x), persediaan_banyak(y))
    z = 7000 - (5000 * alpha)
    return alpha, z

print(
    "Jika di ketahui Permintaan = ", p,
    " dan Persediaan = ", q,
    " maka Produksi = ", tsukamoto(p, q)
) 
```

## 1.2 Logika Fuzzy Mamdani

Contoh 2: Membangun Logika Fuzzy Metode Mamdani

```python
import numpy as np 
import pandas as pd
# Define the fuzzy sets for homework and exam scores 
homework_low = np.array([0, 30, 60]) 
homework_medium = np.array([30, 70, 100]) 
homework_high = np.array([70, 100, 100])
exam_low = np.array([0, 30, 60]) 
exam_medium = np.array([30, 70, 100]) 
exam_high = np.array([70, 100, 100]) 
 
# Define the input values (homework and exam scores) 
homework_score = 80  # For example, homework score of 45 
exam_score = 75     # For example, exam score of 75 
 
# Calculate the membership values for each input in the fuzzy sets 
homework_low_membership = np.interp(homework_score, homework_low, [0, 1, 0]) 
homework_medium_membership = np.interp(homework_score, homework_medium, [0, 1, 
0]) 
homework_high_membership = np.interp(homework_score, homework_high, [0, 1, 0]) 
 
exam_low_membership = np.interp(exam_score, exam_low, [0, 1, 0]) 
exam_medium_membership = np.interp(exam_score, exam_medium, [0, 1, 0]) 
exam_high_membership = np.interp(exam_score, exam_high, [0, 1, 0]) 
 
# Define the fuzzy rules 
rules = [ 
    (min(homework_low_membership, exam_low_membership), "grade_low"), 
    (min(homework_medium_membership, exam_low_membership), "grade_medium"), 
    (min(homework_high_membership, exam_low_membership), "grade_medium"), 
    (min(homework_low_membership, exam_medium_membership), "grade_medium"), 
    (min(homework_medium_membership, exam_medium_membership), "grade_medium"), 
    (min(homework_high_membership, exam_medium_membership), "grade_high"), 
    (min(homework_low_membership, exam_high_membership), "grade_high"), 
    (min(homework_medium_membership, exam_high_membership), "grade_high"), 
    (min(homework_high_membership, exam_high_membership), "grade_high"), 
] 
 
# Aggregate the rule outputs 
grades = { 
    "grade_low": 0, 
    "grade_medium": 0, 
    "grade_high": 0, 
} 
 
for rule in rules: 
    membership, grade = rule 
    grades[grade] = max(grades[grade], membership) 
 
# Calculate the centroid defuzzified grade 
defuzzified_grade = (grades["grade_low"] * 30 + grades["grade_medium"] * 70 + 
grades["grade_high"] * 100) / (grades["grade_low"] + grades["grade_medium"] + 
grades["grade_high"]) 
 
if defuzzified_grade <= 40: 
    output_grade = "Low" 
elif defuzzified_grade <= 70: 
    output_grade = "Medium" 
else: 
    output_grade = "High"

# Output the recommended grade 
print("Recommended Grade:", output_grade) 
```

Modifikasi Contoh 2 untuk menyelesaikan kasus pada modul

```python
def konversi_keanggotaan(nilai):
    if nilai < 60:
        return 1      # Low
    elif nilai < 80:
        return 2      # Middle
    else:
        return 3      # High


def nilai_akhir_fuzzy(tugas, uts, uas):
    # 2.b: rata-rata tugas sebelum dihitung fuzzy
    tugas_rata = sum(tugas) / len(tugas)

    # 2.a: konversi tiap nilai keanggotaan
    total_konversi = (
        konversi_keanggotaan(tugas_rata) +
        konversi_keanggotaan(uts) +
        konversi_keanggotaan(uas)
    )

    # 2.a: hasil fuzzy
    hasil = total_konversi / 3

    # 2.a: kategorisasi hasil fuzzy
    if hasil < 1.5:
        status = "Low"
    elif hasil < 2.5:
        status = "Middle"
    else:
        status = "High"

    # 2.c: konversi ke nilai akademik
    if status == "Low":
        nilai_akademik = "C"
    elif status == "Middle":
        nilai_akademik = "B"
    else:
        nilai_akademik = "A"

    return {
        "rata_tugas": tugas_rata,
        "total_konversi": total_konversi,
        "hasil": hasil,
        "status": status,
        "nilai_akademik": nilai_akademik,
    }


# Kasus Budi
budi = {
    "nama": "Budi",
    "tugas": [80, 85, 75],
    "uts": 80,
    "uas": 40,
}

hasil_budi = nilai_akhir_fuzzy(budi["tugas"], budi["uts"], budi["uas"])
print(f"Nama: {budi['nama']}")
print(f"Rata-rata tugas: {hasil_budi['rata_tugas']}")
print(f"Hasil fuzzy: {hasil_budi['hasil']:.2f} -> {hasil_budi['status']}")
print(f"Nilai akademik: {hasil_budi['nilai_akademik']}")

# Data teman-teman Budi
mahasiswa = [
    {"nama": "Jasmine", "tugas": [70, 80, 75], "uts": 80, "uas": 40},
    {"nama": "Rafik", "tugas": [90, 80, 80], "uts": 77, "uas": 90},
    {"nama": "Mutia", "tugas": [50, 55, 60], "uts": 10, "uas": 30},
    {"nama": "Patrik", "tugas": [10, 0, 90], "uts": 60, "uas": 70},
    {"nama": "Fajar", "tugas": [65, 35, 100], "uts": 70, "uas": 72},
]

hasil = []

for m in mahasiswa:
    hasil_fuzzy = nilai_akhir_fuzzy(m["tugas"], m["uts"], m["uas"])
    hasil.append([
        m["nama"],
        round(hasil_fuzzy["rata_tugas"], 2),
        round(hasil_fuzzy["hasil"], 2),
        hasil_fuzzy["status"],
        hasil_fuzzy["nilai_akademik"]
    ])

data = [
    ("Jasmine", 70, 80, 75, 80, 40),
    ("Rafik", 90, 80, 80, 77, 90),
    ("Mutia", 50, 55, 60, 10, 30),
    ("Patrik", 10, 0, 90, 60, 70),
    ("Fajar", 65, 35, 100, 70, 72),
]

hasil = []

for nama, t1, t2, t3, uts, uas in data:
    tugas_rata = (t1 + t2 + t3) / 3
    total_konversi = (
        konversi_keanggotaan(tugas_rata) + konversi_keanggotaan(uts) + konversi_keanggotaan(uas)
    )
    hasil_fuzzy = total_konversi / 3

    if hasil_fuzzy < 1.5:
        status = "Low"
        nilai_akademik = "C"
    elif hasil_fuzzy < 2.5:
        status = "Middle"
        nilai_akademik = "B"
    else:
        status = "High"
        nilai_akademik = "A"

    hasil.append([
        nama,
        round(tugas_rata, 2),
        round(hasil_fuzzy, 2),
        status,
        nilai_akademik
    ])

df = pd.DataFrame(
    hasil,
    columns=["Nama", "Rata-rata Tugas", "Hasil Fuzzy", "Status", "Nilai Akademik"]
)
df
```

## 1.3 Logika Fuzzy Sugeno

Contoh 3: Membangun Logika Fuzzy Metode Sugeno

```python
import numpy as np 
 
# Define the input values (homework and exam scores) 
homework_score = 50 
exam_score = 50 
 
# Define the fuzzy sets for homework and exam scores 
homework_low = [0, 30, 60] 
homework_medium = [30, 70, 100] 
homework_high = [70, 100, 100] 
 
exam_low = [0, 30, 60] 
exam_medium = [30, 70, 100] 
exam_high = [70, 100, 100] 
 
# Define the input membership function 
def fuzzy_membership(value, low, medium, high): 
    return [ 
        max(0, min((value - low) / (medium - low), (high - value) / (high - 
medium))), 
        max(0, min((value - low) / ((high - low) / 2), 2 - (value - low) / 
((high - low) / 2))), 
        max(0, min(1, (value - medium) / (high - medium))) 
    ] 
 
# Calculate the membership values for the input values 
homework_membership = fuzzy_membership(homework_score, *homework_low) 
exam_membership = fuzzy_membership(exam_score, *exam_low) 
 
# Define the fuzzy rules and their corresponding linear functions 
rules = [ 
    (homework_membership[0], exam_membership[0], lambda x, y: 0.5 * x + 0.5 * y), 
    (homework_membership[0], exam_membership[1], lambda x, y: 0.5 * x + 0.5 * y), 
    (homework_membership[0], exam_membership[2], lambda x, y: 0.5 * x + 0.5 * y), 
    (homework_membership[1], exam_membership[0], lambda x, y: 0.5 * x + 0.5 * y), 
    (homework_membership[1], exam_membership[1], lambda x, y: 0.5 * x + 0.5 * y), 
    (homework_membership[1], exam_membership[2], lambda x, y: 0.5 * x + 0.5 * y), 
    (homework_membership[2], exam_membership[0], lambda x, y: 0.5 * x + 0.5 * y),
    (homework_membership[2], exam_membership[1], lambda x, y: 0.5 * x + 0.5 * y), 
    (homework_membership[2], exam_membership[2], lambda x, y: 0.5 * x + 0.5 * y), 
] 
 
# Calculate the weighted average of the linear functions 
numerator = sum(weight * rule(homework_score, exam_score) for weight, _, rule in 
rules) 
denominator = sum(weight for weight, _, _ in rules) 
 
# Calculate the final result (Sugeno model) 
if denominator != 0: 
    result = numerator / denominator
else: 
    result = 0  # Handle the case where the denominator is 0 
 
if result <= 40: 
    output_grade = "Low" 
elif result <= 70: 
    output_grade = "Medium" 
else: 
    output_grade = "High" 
 
# Output the recommended result (in this case, a grade) 
print("Recommended Grade:", output_grade) 
```

Modifikasi Contoh 3 untuk menyelesaikan kasus pada modul


```python
BIAYA_MIN, BIAYA_MAX = 900_000, 16_600_000
MINTA_MIN, MINTA_MAX = 50, 485
STOK_MIN,  STOK_MAX  = 5, 230
PROD_MIN,  PROD_MAX  = 80, 472

biaya, minta, stok = 6_000_000, 320, 60


#Fungsi keanggotaan linear (turun / naik)
def turun(x, a, b):
    if x <= a: return 1.0
    if x >= b: return 0.0
    return (b - x) / (b - a)

def naik(x, a, b):
    if x <= a: return 0.0
    if x >= b: return 1.0
    return (x - a) / (b - a)


#fuzzifikasi
mu = {
    "biaya": {"RENDAH":  turun(biaya, BIAYA_MIN, BIAYA_MAX),
              "TINGGI":  naik(biaya,  BIAYA_MIN, BIAYA_MAX)},
    "minta": {"TURUN":   turun(minta, MINTA_MIN, MINTA_MAX),
              "NAIK":    naik(minta,  MINTA_MIN, MINTA_MAX)},
    "stok":  {"SEDIKIT": turun(stok,  STOK_MIN,  STOK_MAX),
              "BANYAK":  naik(stok,   STOK_MIN,  STOK_MAX)},
}

print("FUZZIFIKASI")
for var, sets in mu.items():
    for name, val in sets.items():
        print(f"  mu_{var}_{name:<8} = {val:.4f}")

#rules (Sugeno orde-1: z = f(permintaan, stok))
rules = [
    # (biaya,   permintaan, stok,     rumus z,                 fungsi z)
    ("RENDAH", "TURUN", "BANYAK",  "Permintaan - Stok",        lambda p, s: p - s),
    ("RENDAH", "TURUN", "SEDIKIT", "Permintaan",               lambda p, s: p),
    ("RENDAH", "NAIK",  "BANYAK",  "Permintaan",               lambda p, s: p),
    ("RENDAH", "NAIK",  "SEDIKIT", "1,25*Permintaan - Stok",   lambda p, s: 1.25 * p - s),
    ("TINGGI", "TURUN", "BANYAK",  "Permintaan - Stok",        lambda p, s: p - s),
    ("TINGGI", "TURUN", "SEDIKIT", "Permintaan - Stok",        lambda p, s: p - s),
    ("TINGGI", "NAIK",  "BANYAK",  "Permintaan - Stok",        lambda p, s: p - s),
    ("TINGGI", "NAIK",  "SEDIKIT", "Permintaan",               lambda p, s: p),
]

#inferensi (MIN) + Defuzzifikasi (rata-rata terbobot)
print("\nINFERENSI")
num = den = 0.0
for i, (b, p, s, rumus, f) in enumerate(rules, 1):
    alpha = min(mu["biaya"][b], mu["minta"][p], mu["stok"][s])
    z = min(max(f(minta, stok), PROD_MIN), PROD_MAX)   # jaga di [80, 472]
    num += alpha * z
    den += alpha
    print(f"  R{i}: IF Biaya {b:<6} AND Permintaan {p:<5} AND Stok {s:<7} "
          f"THEN Produksi = {rumus:<23} -> a={alpha:.4f}, z={z:.2f}")

Z = num / den if den else 0
print("\nDEFUZZIFIKASI")
print(f"  Z = sum(a*z)/sum(a) = {num:.4f} / {den:.4f} = {Z:.2f}")
print(f"  Jumlah kain tenun yang diproduksi Maret 2017 ~ {round(Z)} biji")
```
